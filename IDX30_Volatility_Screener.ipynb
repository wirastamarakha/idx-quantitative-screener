```python
# Install library yfinance (wajib dijalankan pertama kali di Colab)
!pip install yfinance

import yfinance as yf
import pandas as pd

# Kode emiten pasar modal Indonesia di Yahoo Finance wajib menggunakan akhiran ".JK"
ticker_saham = ['AMMN.JK', 'ANTM.JK', 'ARTO.JK', 'ASII.JK', 'BBCA.JK',
    'BBNI.JK', 'BBRI.JK', 'BMRI.JK', 'BRPT.JK', 'CPIN.JK',
    'EMTK.JK', 'ESSA.JK', 'EXCL.JK', 'GOTO.JK', 'HRUM.JK',
    'ICBP.JK', 'INCO.JK', 'INDF.JK', 'INKP.JK', 'ITMG.JK',
    'KLBF.JK', 'MDKA.JK', 'MEDC.JK', 'PGAS.JK', 'PGEO.JK',
    'PTBA.JK', 'SMGR.JK', 'TLKM.JK', 'TOWR.JK', 'UNVR.JK']

# Menarik data harga penutupan (Close) secara otomatis dari awal tahun 2026
print("Mengakses server Yahoo Finance...")
data_harga = yf.download(ticker_saham, start='2026-01-01')['Close']

# Menampilkan 5 hari perdagangan terakhir
print("\nData 5 Hari Terakhir:")
display(data_harga.tail())
```

```python
import numpy as np

# 1. Menghitung persentase keuntungan/kerugian harian (Daily Returns)
daily_returns = data_harga.pct_change().dropna()

# 2. Menghitung Standar Deviasi (Risiko) lalu disetahunkan (dikali akar 252 hari bursa)
annual_volatility = daily_returns.std() * np.sqrt(252)

# 3. Mengurutkan dari saham paling tidak berisiko sampai paling berisiko (agresif)
sorted_volatility = annual_volatility.sort_values()

print("Tingkat Volatilitas Tahunan (Risiko) Emiten:")
print("-" * 45)
print(sorted_volatility.map("{:.2%}".format))
print("-" * 45)
print("*Semakin tinggi persentasenya, semakin liar pergerakan harganya.")
```

```python
from google.colab import files

# Menyusun data ke dalam format tabel (DataFrame) yang rapi untuk Excel
df_volatility = pd.DataFrame(sorted_volatility, columns=['Volatilitas Tahunan'])

# Mengubah format angka menjadi persentase agar langsung terbaca rapi di Excel
df_volatility['Volatilitas Tahunan'] = df_volatility['Volatilitas Tahunan'].apply(lambda x: f"{x:.2%}")

# Memberikan label keterangan sederhana
df_volatility.index.name = 'Kode Emiten'

# Mengekspor ke file Excel
nama_file = 'IDX_Volatility_Report.xlsx'
df_volatility.to_excel(nama_file, sheet_name='Analisis Risiko')

print(f"Mengekspor data ke {nama_file}...")

# Memunculkan pop-up download otomatis ke laptop
files.download(nama_file)
```
